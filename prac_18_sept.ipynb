Method overloading

```python
class Calculator:
    def add(self, a, b=0):
        return a + b
obj = Calculator()
print(obj.add(10))
print(obj.add(10, 20))
```

```python
class Calculator:
    def add(self, *numbers):
        total = 0
        for num in numbers:
            total += num
        return total
obj = Calculator()
print(obj.add(10, 20))
print(obj.add(10, 20, 30, 40))
```

method overriding

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
obj = Dog()
obj.sound()
```

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")
class Car(Vehicle):
    def start(self):
        print("Car starts with key")
c = Car()
c.start()
```

class polymorphism

```python
class Dog:
    def sound(self):
        print("Dog barks")
class Cat:
    def sound(self):
        print("Cat meows")
def make_sound(animal):
    animal.sound()
d = Dog()
c = Cat()
make_sound(d)
make_sound(c)
```

```python
class Circle:
    def draw(self):
        print("Drawing Circle")
class Rectangle:
    def draw(self):
        print("Drawing Rectangle")
def display(shape):
    shape.draw()
c = Circle()
r = Rectangle()
display(c)
display(r)
```

abstraction

```python
class Vehicle:
    def start(self):
        print("Start method should be implemented by child class")
class Car(Vehicle):
    def start(self):
        print("Car starts with key")
obj = Car()
obj.start()
```

```python
class Shape:
    def area(self):
        print("Area method should be implemented by child class")
class Circle(Shape):
    def area(self):
        print("Area of Circle = 3.14 × r × r")
obj = Circle()
obj.area()
```
