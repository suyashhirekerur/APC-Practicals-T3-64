#Constructor and Destructor in a Class

```python
# Write a Python program to demonstrate the use of a constructor and destructor in a Student class.
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
        print("Constructor called")
        print("Student created:", self.name)
    def __del__(self):
        print("Destructor called")
        print("Student object deleted:", self.name)
student1 = Student("Suyash", 64)
print("Name:", student1.name)
print("Roll No:", student1.roll_no)
del student1
```

```python
# Write a Python program to demonstrate constructor and destructor in a BankAccount class.
class BankAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance
        print("Constructor called")
        print("Account created:", self.account_no)
    def display(self):
        print("Account Number:", self.account_no)
        print("Balance:", self.balance)
    def __del__(self):
        print("Destructor called")
        print("Account closed:", self.account_no)
account = BankAccount(12345, 50000)
account.display()
del account
```

```python
# Write a Python program to demonstrate the working of a constructor and destructor using an Employee class.
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        print("Constructor executed")
    def display(self):
        print("Employee Name:", self.name)
        print("Salary:", self.salary)
    def __del__(self):
        print("Destructor executed for", self.name)
emp = Employee("Rahul", 40000)
emp.display()
del emp
```

#Types of Inheritance

A. Single Inheritance

```python
# Write a Python program to demonstrate single inheritance using Person and Student classes.
class Person:
    def display_person(self):
        print("I am a person")
class Student(Person):
    def display_student(self):
        print("I am a student")
student = Student()
student.display_person()
student.display_student()
```

```python
# Write a Python program to demonstrate multiple inheritance using Teacher, Researcher, and Professor classes.
class Teacher:
    def teaching(self):
        print("Teacher teaches students")
class Researcher:
    def research(self):
        print("Researcher performs research")
class Professor(Teacher, Researcher):
    def guide(self):
        print("Professor guides students")
professor = Professor()
professor.teaching()
professor.research()
professor.guide()
```

B. Multiple Inheritance

```python
# Write a Python program to demonstrate multiple inheritance using Father, Mother, and Child classes.
class Father:
    def father_property(self):
        print("Father's property")
class Mother:
    def mother_property(self):
        print("Mother's property")
class Child(Father, Mother):
    def child_property(self):
        print("Child's property")
child = Child()
child.father_property()
child.mother_property()
child.child_property()
```

```python
# Write a Python program to demonstrate multiple inheritance using Teacher, Researcher, and Professor classes.
class Teacher:
    def teaching(self):
        print("Teacher teaches students")
class Researcher:
    def research(self):
        print("Researcher performs research")
class Professor(Teacher, Researcher):
    def guide(self):
        print("Professor guides students")
professor = Professor()
professor.teaching()
professor.research()
professor.guide()
```

C. Multilevel Inheritance

```python
# Write a Python program to demonstrate multilevel inheritance using Animal, Mammal, and Dog classes.
class Animal:
    def eat(self):
        print("Animal eats food")
class Mammal(Animal):
    def walk(self):
        print("Mammal walks")
class Dog(Mammal):
    def bark(self):
        print("Dog barks")
dog = Dog()
dog.eat()
dog.walk()
dog.bark()
```

D. Hierarchical Inheritance

```python
# Write a Python program to demonstrate hierarchical inheritance using Shape as the base class and Circle and Rectangle as derived classes.
class Shape:
    def display(self):
        print("This is a shape")
class Circle(Shape):
    def circle_area(self, radius):
        print("Circle Area:", 3.14 * radius * radius)
class Rectangle(Shape):
    def rectangle_area(self, length, breadth):
        print("Rectangle Area:", length * breadth)
circle = Circle()
rectangle = Rectangle()
circle.display()
circle.circle_area(5)
rectangle.display()
rectangle.rectangle_area(10, 5)
```

E. Hybrid Inheritance

```python
# Write a Python program to demonstrate hybrid inheritance using Person, Student, Employee, and Intern classes.
class Person:
    def display_person(self):
        print("I am a person")


class Student(Person):
    def study(self):
        print("Student is studying")


class Employee(Person):
    def work(self):
        print("Employee is working")


class Intern(Student, Employee):
    def training(self):
        print("Intern is undergoing training")


intern = Intern()

intern.display_person()
intern.study()
intern.work()
intern.training()
```

#Private and Protected Variables/Functions in Multilevel Inheritance

```python
# Write a Python program to demonstrate private variables and private functions in a multilevel inheritance hierarchy.
class Person:
    def __init__(self):
        self.__name = "Suyash"
    def __display_name(self):
        print("Name:", self.__name)
    def show_person(self):
        self.__display_name()
class Employee(Person):
    def show_employee(self):
        print("Employee class")
class Manager(Employee):
    def show_manager(self):
        print("Manager class")
manager = Manager()
manager.show_person()
manager.show_employee()
manager.show_manager()
```

```python
# Write a Python program to demonstrate both protected and private members in multilevel inheritance using Person, Employee, and Manager classes.
class Person:
    def __init__(self):
        self._name = "Suyash"       # Protected
        self.__age = 20             # Private
    def _display_name(self):        # Protected function
        print("Name:", self._name)
    def __display_age(self):        # Private function
        print("Age:", self.__age)
    def display_person(self):
        self._display_name()
        self.__display_age()
class Employee(Person):
    def employee_details(self):
        print("Employee Details")
        print("Protected Name:", self._name)
        self._display_name()
class Manager(Employee):
    def manager_details(self):
        print("Manager Details")
        print("Protected Name:", self._name)
        self._display_name()
manager = Manager()
manager.display_person()
manager.employee_details()
manager.manager_details()
```

```python
class BankAccount:
    account_counter = 1000
    def __init__(self, account_holder):
        self.account_holder = account_holder
        self.balance = 0
        BankAccount.account_counter += 1
        self.account_number = BankAccount.account_counter
    def deposit(self, amount):
        self.balance += amount
        print(f"₹{amount} deposited into {self.account_holder}'s account.")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"₹{amount} withdrawn from {self.account_holder}'s account.")
        else:
            print("Error: Insufficient funds.")
    def display_balance(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Account Number: {self.account_number}")
        print(f"Current Balance: ₹{self.balance}")
    def transfer(self, amount, other_account):
        if amount <= self.balance:
            self.balance -= amount
            other_account.balance += amount
            print(
                f"₹{amount} transferred from "
                f"{self.account_holder} to {other_account.account_holder}."
            )
        else:
            print("Error: Insufficient funds for transfer.")
account1 = BankAccount("Suyash")
account2 = BankAccount("Jagdish")
account1.deposit(10000)
account2.deposit(5000)
print("\n--- Initial Balances ---")
account1.display_balance()
print()
account2.display_balance()
print("\n--- Withdrawal ---")
account1.withdraw(2000)
print("\n--- Transfer ---")
account1.transfer(3000, account2)
print("\n--- Final Balances ---")
account1.display_balance()
print()
account2.display_balance()
```
