# Word Tokenization

```python
# basic word tokenization
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
text = "Python is easy to learn."
tokens = word_tokenize(text)
print(tokens)
```

```python
# word tokenization using user input
from nltk.tokenize import word_tokenize
text = input("Enter a sentence: ")
tokens = word_tokenize(text)
print("Words:", tokens)
print("Number of tokens:", len(tokens))
```

```python
# count words
from nltk.tokenize import word_tokenize
text = "Python is powerful and Python is easy."
tokens = word_tokenize(text)
words = [word.lower() for word in tokens if word.isalpha()]
print("Tokens:", words)
print("Total words:", len(words))
```

# class with function

```python
# basic class and function
class Student:
  def display(self):
    print("Hello, I am a student.")
s1 = Student()
s1.display()
```

```python
# class function with parameters
class Student:
    def display(self, name, age):
        print("Name:", name)
        print("Age:", age)

s1 = Student()
s1.display("Suyash", 20)
```

```python
# class with multiple functions
class Calc:
  def add(self, a, b):
    return a + b
  def sub(self, a, b):
    return a - b
  def mul(self, a, b):
    return a * b
c = Calc()
print("Addition:", c.add(10, 20))
print("Subtraction:", c.sub(10, 20))
print("Multiplication:", c.mul(10, 20))
```

#Constructor

```python
# basic constructor
class Student:
    def __init__(self):
        print("Constructor called")
        print("Student object created")
s1 = Student()
```

```python
# constructor with parameters
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def display(self):
      print("Name: ", self.name)
      print("Age: ", self.age)
s1 = Student("Suyash", 21)
s1.display()
```

```python
# employee constructor
class Employee:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary
  def display(self):
    print("name: ", self.name)
    print("salary: ", self.salary)
e1 = Employee("Suyash", 10000)
e1.display()
```

# Destructor

```python
# destructor with parameters
class Student:
  def __init__(self):
    print("Object Created")
  def __del__(self):
    print("Object Destroyed")
s1 = Student()
del s1
```

```python
# cons. + dests.
class Demo:
    def __init__(self):
        print("Constructor executed")
    def __del__(self):
        print("Destructor executed")
obj = Demo()
print("Program is running...")
del obj
```

```python
# destuctor with multiple objects
class Student:
    def __init__(self, name):
        self.name = name
        print(self.name, "object created")
    def __del__(self):
        print(self.name, "object destroyed")
s1 = Student("Suyash")
s2 = Student("Rahul")
del s1
del s2
```

#Sentence Tokenization

```python
# basic sentence tokenization
from nltk.tokenize import sent_tokenize
text = "Python is an easy to learn programming language. Python is powerful and versatile."
sentences = sent_tokenize(text)
print(sentences)
```

```python
# using user input
from nltk.tokenize import sent_tokenize
text = input("Enter some sentences: ")
sentences = sent_tokenize(text)
print("Sentences: ", sentences)
print("Number of sentences: ", len(sentences))
```

```python
# count sentence
from nltk.tokenize import sent_tokenize
text = """Python is a popular programming language.

Python is easy to understand and learn.

Python is used to create websites and applications.

Python is commonly used in Artificial Intelligence.

Python is useful for Data Science and data analysis.

Python has many built-in functions and libraries.

Python allows programmers to write simple and clean code.

Python is used to automate many tasks.

Python is a high-level and powerful programming language.

Python is used by beginners as well as professional developers.

Python supports object-oriented programming.

Python can be used to develop games and software.

Python has a large community of developers.
"""
sentences = sent_tokenize(text)
print("Sentences: ")
for i, sentence in enumerate(sentences, start=1):
    print(i, ":", sentence)
print("Total sentences: ", len(sentences))
```

#protected member using _

```python
class Student:
  def __init__(self, name, marks):
    self.name = name
    self._marks = marks
  def display(self):
    print("Name: ", self.name)
    print("Marks: ", self._marks)
s1 = Student("Suyash", 90)
s1.display()
print(f"Protected marks: {s1._marks}")
```
